```python
import random
nix=True

#Items
Dynitem=0
Kill_coun=[0]
itemlist=[2,2,2,2,2,3]
itemnmlist=['old Food','Wpn','Armor','PMD','Artifact','Gold']

PkPlayeralist=[12,8,4,8,8,12]#Player
DauPlayeralist=[10,3,1,6,2,0]
StaPlayeralist=[10,3,1,6,2,0]
DynPlayeralist=[10,3,1,6,2,0]
TexPlayeralist=['Hp','Atk','Def','Spd','Mp','Gp','옷']
Slimelist=['slime',4,1,0,1,1,1,0,'(*)']#Slime
Ratlist=['rat',6,1,1,3,1,0,3,'<,=,~']#Rat
Spiderlist=['spider',6,2,1,4,2,1,2,'oM0`']#Spider
Ghoullist=['ghoul',7,3,1,4,1,2,0,'옷`']#Ghoul
Golemlist=['golem',10,4,2,2,2,0,1,';⏍;']#Golem

Dynene=[0]#current enemy
monstatlist=[Slimelist,Ratlist,Spiderlist,Ghoullist,Golemlist]
current_ene=[0]
Trap_val=[0]
#enemy_stats=['Name','Hp','Atk','Def','Spd','Off','dev','Crit','Spr']

def Display(Map:int):
    match Map:
        case 0:
            print('==================================================')
            print('')
            print(F'  {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 1:
            print('==================================================')
            print('')
            print(F'        {TexPlayeralist[6]}     [+[-]')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 2:
            print('|||||* *|||||||||||||/============================')
            print('|||||||||||* *|||||/')
            print(F'||||* *||||||||||/  {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 3:
            match Trap_val[0]:
                case 1:
                    print('==================================================')
                    print('')
                    print('        >+O')
                    print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
                case 2:
                    print('==================================================')
                    print('      ')
                    print('      ')
                    print(F'mmmmmmmmmm {TexPlayeralist[6]} mmmmmmmmmmmmmmmmmmmmmmmmmmmm')
                case 3:
                    print('==================================================')
                    print('')
                    print('')
                    print('mmmmmmmmmm-┓      O+<')
                    print('mmmmmmmmmmmm-┓')
                    print('mmmmmmmmmmmmmm-┓')
                    print('mmmmmmmmmmmmmmmm-┓')
        case 4:
            print('==================================================')
            print('')
            print(F'      {TexPlayeralist[6]}              {current_ene[7]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 5:
            print('=== === ==========================================')
            print('   ^   ^')
            print(F'           {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 6:
            print('==================================================')
            print('      [-|-|-]  [-|-|-]')
            print(F'      [-|-|-]  [-|-|-]  {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 7:
            print('==================================================')
            print('                   #     ((+))      #')
            print(F'    {TexPlayeralist[6]}    ','+      -|-       +')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 8:
            print('==================================================')
            print('')
            print(F'                   {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')
        case 9:
            print('==================================================')
            print('      ???')
            print(F'     ?????        {TexPlayeralist[6]}')
            print('mmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmmm')

def Perm_pla_Stats(a:list,b:list,Hp:int,Atk:int):
    a[Atk]=Hp
    b[Atk]=Hp

def Pl_pla_Stats():
    print(" ")
    print("Player Stats:")
    print(F'Hp: {DynPlayeralist[0]}//Atk: {DynPlayeralist[1]}//Def: {DynPlayeralist[2]}')
    print(F'Spd: {DynPlayeralist[3]}//Mp: {DynPlayeralist[4]}//Gp: {DynPlayeralist[5]}')
    print(" ")

def is_pladead(hp:int):
    if hp<=0:
        print(" ")
        print("you Died! Rip.")
    return(hp)

def Game_RNG():
    num=random.randint(1-int(DynPlayeralist[5]/2),6)
    match num:
        case 6:
            Chest()
        case 5:
            Discription()
        case 2:
            trap()
    if num<=1:
        if Kill_coun[0]>=10:
            Discription()
        else:
            Encounter()
    elif num==4 or num==3:
        Explore()

def trap():
    victim=random.randint(0, 6)
    if victim>=5:
        victim=5
        if DynPlayeralist[5]==0:
            victim=random.randint(0, 4)
    num=random.randint(1,3)
    n=num
    if victim==0:
        n=n-StaPlayeralist[2]
    print(" ")
    print(" ")
    Trap_val[0]=num
    match num:
        case 1:
            print(f'you tripped and fell, Losing {n} {TexPlayeralist[victim]}')
        case 2:
            print(f'you fell into a Pit, Losing {n} {TexPlayeralist[victim]}')
        case 3:
            print(f'you fell down a flight of Stairs, Losing {n} {TexPlayeralist[victim]}')
    if n<0:
        n=0
    Display(3)
    n=StaPlayeralist[victim]-n
    Perm_pla_Stats(StaPlayeralist,DynPlayeralist,n,victim)
    perm_hurt(0)

def perm_hurt(n):
    p=StaPlayeralist[0]-n
    Perm_pla_Stats(StaPlayeralist,DynPlayeralist,p,0)
    Pl_pla_Stats()
    if is_pladead(DynPlayeralist[0])>0:
        while nix:
            try:
                print(" ")
                print("Input 2 to recontinue exploring the overgrown Dungean ")
                print("Input 0 to leave the Dungeon ")
                age=int(input("Input Here ->: "))
                if age==2:
                    Explore()
                    break
                elif age==0:
                    Dungean()
                    break
            except ValueError:
                print("Invalid input!")

def Encounter():
    num=random.randint(0,4)
    print(" ")
    print(" ")
    Dynene[0]=num
    print(F'there is a {monstatlist[Dynene[0]][0]}')
    current_ene[:]=monstatlist[num][1:9]
    Display(4)
    while nix:
        try:
            print('')
            print('Input 2 to engage')
            print('Input 1 to avoid')
            print('Input 0 to leave the Dungeon')
            sol=int(input("Input Here ->: "))
            match sol:
                case 2:
                    combat(False)
                    break
                case 1:
                    S_cape()
                    break
                case 0:
                    Dungean()
                    break
        except ValueError:
            print("Invalid input!")

def S_cape():
    no=random.randint(current_ene[3],StaPlayeralist[3])
    if no>=StaPlayeralist[3] and current_ene[0]>0:
        n=current_ene[1]-StaPlayeralist[2]
        print(F'you try to avoid the {monstatlist[Dynene[0]][0]} but it nicked you as you fled. taking {n} Dmg!')
        perm_hurt(n)
    else:
        print(F'you Successfully avoided the {monstatlist[Dynene[0]][0]}!')
        perm_hurt(0)

def combat(dev:bool):
    has_act=True
    has_defend=False
    if current_ene[0]>0:
        while nix:
            try:
                Display(4)
                Pl_pla_Stats()
                print('Input 3 to use magic')
                print('Input 2 to Atk')
                print('Input 1 to Defend')
                print('Input 0 to leave the Fight')
                sol=int(input("Input Here ->: "))
                match sol:
                    case 3:
                        print(' ')
                        print('Input 1 to Heal Cost: 1 Mp')
                        print('Input 2 to Slow Enemy Cost: 2 Mp')
                        print('Input 3 to Rust enemy Atk or Def Cost: 3 Mp')
                        print('Input 0 to Return to main options')
                        age=int(input("Input Here ->: "))
                        Magic(age)
                        has_act=False
                        break
                    case 2:
                        imp=random.randint(-1,8)
                        if imp>current_ene[3] or current_ene[3]==0:
                            if imp==DynPlayeralist[3]:
                                current_ene[0]=current_ene[0]-((DynPlayeralist[1]*2)-current_ene[2])
                                print(F'you Critted the {monstatlist[Dynene[0]][0]}! dealing: {(DynPlayeralist[1]*2)-current_ene[2]} Dmg')
                            else:
                                if dev==False:
                                    current_ene[0]=current_ene[0]-(DynPlayeralist[1]-current_ene[2])
                                    print(F'you Hit the {monstatlist[Dynene[0]][0]}! dealing: {DynPlayeralist[1]-current_ene[2]} Dmg')
                                else:
                                    print(F'the {monstatlist[Dynene[0]][0]} blocked your attack!')
                        else:
                            if imp<0:
                                imp=abs(imp)
                            print(F'You Missed! by: {imp+1}ft')
                        if current_ene[0]<=0:
                            kill_cap()
                            has_act=False
                        break
                    case 1:
                        print('The You chose to Defend')
                        has_defend=True
                        break
                    case 0:
                        has_act=False
                        if current_ene[0]>0:
                            imp=random.randint(current_ene[3]-1,DynPlayeralist[3])
                            if imp<=current_ene[3]:
                                has_act=True
                                print(F'The {monstatlist[Dynene[0]][0]} Chased You!')
                            else:
                                S_cape()
                        else:
                            S_cape()
                        break
            except ValueError:
                print("Invalid input!")
        if has_act==True:
            ene_combat(has_defend)
def ene_combat(dev:bool):
    cob=random.randint(0-current_ene[5], 1+current_ene[6])
    if cob<=0:
        imp=random.randint(-1,8)
        if imp<=current_ene[3]or current_ene[3]==0:
            if imp==DynPlayeralist[3]:
                dmg=DynPlayeralist[0]-((current_ene[1]*2)-DynPlayeralist[2])
                if dmg<0:
                    dmg=0
                pri=F'Dealing: {(current_ene[1]*2)-DynPlayeralist[2]} Dmg'
                Crit_Ability=current_ene[7]
                trig=random.randint(0,1)
                if trig<1:
                    Crit_Ability=0
                else:
                    if Crit_Ability>=1:
                        pri=' '
                print(F'The {monstatlist[Dynene[0]][0]} Critted You!',pri)
                match Crit_Ability:
                    case 1:
                        dmg=DynPlayeralist[0]-(current_ene[1]*2)
                        print(F'The {monstatlist[Dynene[0]][0]} pierced You! Dealing: {(current_ene[1]*2)} Dmg')
                    case 2:
                        DynPlayeralist[3]=DynPlayeralist[3]-1
                        print(F'The {monstatlist[Dynene[0]][0]} Slowed You! by: {-1}')
                    case 3:
                        DynPlayeralist[1]=DynPlayeralist[1]-1
                        print(F'The {monstatlist[Dynene[0]][0]} Impaired You! by: {-1}')
                DynPlayeralist[0]=dmg
            else:
                if dev==False:
                    dmg=DynPlayeralist[0]-(current_ene[1]-DynPlayeralist[2])
                    if dmg<0:
                        dmg=0
                    DynPlayeralist[0]=dmg
                    print(F'the {monstatlist[Dynene[0]][0]} Hit you! Dealing: {(current_ene[1])-DynPlayeralist[2]} Dmg')
                else:
                    print(F'You blocked the {monstatlist[Dynene[0]][0]} attack!')
        else:
            if imp<0:
                imp=abs(imp)
            print(F'The {monstatlist[Dynene[0]][0]} Missed! by: {imp}ft')
        if DynPlayeralist[0]<=0:
            is_pladead(DynPlayeralist[0])
        else:
            combat(False)
    elif cob>=1:
        print(F'The {monstatlist[Dynene[0]][0]} chose to Defend')
        combat(True)
def Magic(sol:int):
    match sol:
        case 1:
            if DynPlayeralist[4]>0:
                imp=random.randint(2,DynPlayeralist[4])
                DynPlayeralist[4]=DynPlayeralist[4]-1
                DynPlayeralist[0]=DynPlayeralist[0]+imp
                print(F'you healed {imp} Hp')
                if DynPlayeralist[0]>StaPlayeralist[0]:
                    Perm_pla_Stats(DynPlayeralist,DynPlayeralist,StaPlayeralist[0],0)
                ene_combat(False)
            else:
                print('Sorry! you dont have enough Mp to cast this spell')
                combat(False)
        case 2:
            if DynPlayeralist[4]>1:
                imp=random.randint(current_ene[3]-1,DynPlayeralist[3])
                if imp>current_ene[3] or current_ene[3]==0:
                    DynPlayeralist[4]=DynPlayeralist[4]-2
                    current_ene[3]=current_ene[3]-2
                    print(F'you slowed the {monstatlist[Dynene[0]][0]}!')
                    if current_ene[3]<0:
                        current_ene[3]=0
                    else:
                        print(F'You missed! by {imp}ft')
                    ene_combat(False)
            else:
                print('Sorry! you dont have enough Mp to cast this spell')
                combat(False)
        case 3:
            if DynPlayeralist[4]>1:
                imp=random.randint(current_ene[3]-1,DynPlayeralist[3])
                pmi=random.randint(0,1)
                eff=random.randint(0,2)
                if pmi==0:
                    if imp>current_ene[3] or current_ene[3]==0:
                        DynPlayeralist[4]=DynPlayeralist[4]-3
                        current_ene[1]=current_ene[1]-(1+eff)
                        print(F'you rusted the {monstatlist[Dynene[0]][0]} Atk! By {(1+eff)}')
                        if current_ene[1]<1:
                            current_ene[1]=1
                        else:
                            print(F'You missed! by {imp}ft')
                        ene_combat(False)
                else:
                    if imp>current_ene[3] or current_ene[3]==0:
                        DynPlayeralist[4]=DynPlayeralist[4]-3
                        current_ene[2]=current_ene[2]-(1+eff)
                        print(F'you Rusted the {monstatlist[Dynene[0]][0]} Def! By {(1+eff)}')
                        if current_ene[2]<0:
                            current_ene[2]=0
                        else:
                            print(F'You missed! by {imp}ft')
                        ene_combat(False)
            else:
                print('Sorry! you dont have enough Mp to cast this spell')
                combat(False)
        case 0:
            combat(False)

def kill_cap():
    print(F'You did it! Killed the {monstatlist[Dynene[0]][0]}.')
    Kill_coun[0]=Kill_coun[0]+1
    print(F'Total Kills: {Kill_coun[0]}')
    if Kill_coun[0]>=10:
        print('you beat the dungeon!')
    if StaPlayeralist!=DynPlayeralist:
        DynPlayeralist[:]=StaPlayeralist
        print('Stat Reset!')
    perm_hurt(0)
def Chest():
    num=random.randint(0,5)
    print(" ")
    print("on your exploration of these abandend halls you found a chest. what do you do?")
    Display(1)
    while nix:
        try:
            print(" ")
            print("Input 2 to open ")
            print("Input 1 to Skip ")
            print("Input 0 to leave the Dungeon ")
            sol=int(input("Input Here ->: "))
            print(" ")
            match sol:
                case 2:
                    print(F'you opened the chest and found, a {itemnmlist[num]}')
                    open_chest(num)
                    break
                case 1:
                    print(" ")
                    print("you Leave the Rustic chest behind")
                    Explore()
                    break
                case 0:
                    Dungean()
                    break
        except ValueError:
            print("Invalid input!")
def open_chest(item:int):
    i=0
    Display(1)
    while nix:
        try:
            if item==0:
                print(F'Input 2 to Eat the {itemnmlist[item]}')
            else:
                print(F'Input 2 to take the {itemnmlist[item]}')
            print(F'Input 1 to leave the {itemnmlist[item]}')
            print('Input 0 to leave the Dungeon')
            sol=int(input("Input Here ->: "))
            match sol:
                case 2:
                    print(" ")
                    print(F'you took the {itemnmlist[item]} with you')
                    Dynitem=itemlist[item]
                    if item==0:
                        i=DauPlayeralist[0]
                    elif item==1 or item==3 or item==4:
                        i=PkPlayeralist[1]
                    elif item==2:
                        i=PkPlayeralist[4]
                    else:
                        i=PkPlayeralist[0]
                    go=random.randint(1,Dynitem)
                    Dynitem=go
                    if (DynPlayeralist[item]+Dynitem)>i:
                        print('OverFlow')
                    else:
                        print("+",Dynitem,TexPlayeralist[item])
                        i=DynPlayeralist[item]+Dynitem
                    Perm_pla_Stats(StaPlayeralist,DynPlayeralist,i,item)
                    Pl_pla_Stats()
                    Explore()
                    break
                case 1:
                    print(" ")
                    print("you Leave the item behind")
                    Explore()
                    break
                case 0:
                    Dungean()
                    break
        except ValueError:
            print("Invalid input!")

def Discription():
    num=random.randint(1,6)
    print(" ")
    nim=3
    match num:
        case 1:
            print("as you venture through these ForSaken halls you feel dead eyes staring at you in the Dark")
        case 2:
            print("as you Walk down the eroded Stairs you Hears Whispers and crys of the Damned as you walked pass cracked Whispering pipes")
        case 3:
            print("as you venture Deeper into this dungeon you notice the cells that pattern these halls become further blackened and Steined")
        case 4:
            print("as you venture through these WellKnown halls you find yourself in a Forgotten Garden well overgrown and uncemped with a hauntingly beautyful emerald tree at its center")
        case 5:
            print("as you venture through these War Striken wall you feel as if you belong")
        case 6:
            print("as you venture down these Anchent halls you walk pass a sight you wish you Shouldent have Seen")
    if num==1:
        nim=4
    Display(num+nim)
    Explore()

def Explore():
    while nix:
        try:
            Display(0)
            print(" ")
            print('Input 1 to continue into the old Dungeon')
            print('Input 0 to leave the Dungeon')
            age = int(input("Input Here ->: "))
            if age==1:
                Game_RNG()
                break
            elif age==0:
                Dungean()
                break
        except ValueError:
            print("Invalid input!")

def Dungean():
    print(" ")
    print("   o===============o")
    for i in range(12):
        print("   H               H")
    print("mmmmmmmmmmmmmmmmmmmmmmmm")
    print(" ")
    print("welcome to the Dungean")
    while nix:
        try:
            print('Input 2 to enter the Dungeon')
            print('Input 0 to leave the Dungeon')
            age=int(input("Input Here ->: "))
            if age==2:
                print(" ")
                print("you Walked in to the old Dungean")
                Pl_pla_Stats()
                Explore()
                break
            elif age==0:
                start()
                break
        except ValueError:
            print("Invalid input!")

def start():
    while True:
        try:
            print(" ")
            print('Input 1 to start Game')
            print('Input 0 to Quit')
            age = int(input("Input Here ->: "))
            if age==1:
                Dungean()
                break
            if age==0:
                break
        except ValueError:
            print("Invalid input!")
            print(" ")

start()
```
